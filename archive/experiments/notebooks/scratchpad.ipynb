```python
import sys
import os
import asyncio

# Добавляем корень проекта в пути импорта
sys.path.append(os.path.abspath(".."))

from src.models.huggingface_client import generate_summary

async def test_run():
    # Загрузка системного промпта из файла в корне проекта
    prompt_path = "../промпт для модели.txt"
    if os.path.exists(prompt_path):
        with open(prompt_path, "r", encoding="utf-8") as f:
            system_prompt = f.read()
    else:
        system_prompt = "Ты — Senior Data Engineer и BI Architect. Сделай саммари."
    
    # Имитация переписки рабочего чата (mock data)
    mock_chat_history = (
        "Разработчик 1: Обнаружили падение производительности на проде. Запросы к PostgreSQL висят.\n"
        "Аналитик: Я проверил дашборды в DataLens, графики загрузки CPU базы данных на 100%.\n"
        "Разработчик 2: Похоже, забыли накат индекса на таблицу staging_events по полю event_time.\n"
        "Разработчик 1: Сейчас подготовлю миграцию и закрою задачу."
    )
    
    # Объединяем системный промпт и историю сообщений в моно-промпт
    full_content = f"{system_prompt}\n\nИстория сообщений для анализа:\n{mock_chat_history}"
    
    print("Отправка запроса в Hugging Face API...")
    try:
        summary = await generate_summary(full_content)
        print("\n=== РЕЗУЛЬТАТ САММАРИ ===")
        print(summary)
    except Exception as e:
        print(f"Ошибка при тестировании: {e}")
        raise e  # Не скрываем провал теста

# Запуск асинхронной функции
await test_run()
```
